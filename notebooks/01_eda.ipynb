# Phase 1 — FER2013 EDA

Class balance, sample faces, and image properties for FER2013.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

if Path.cwd().name == "notebooks":
    sys.path.insert(0, str(Path.cwd().parent))

from config import CLASS_NAMES, FER_DIR, FIGURES_DIR
from src.utils import ensure_dirs

ensure_dirs()
```

## Class counts

```python
def count_images(split_dir: Path) -> dict:
    counts = {}
    for cls in CLASS_NAMES:
        cls_dir = split_dir / cls
        counts[cls] = len(list(cls_dir.glob("*"))) if cls_dir.exists() else 0
    return counts


train_counts = count_images(FER_DIR / "train")
test_counts = count_images(FER_DIR / "test")
print("Train:", train_counts)
print("Test:", test_counts)
```

## Class distribution bar chart

```python
fig, ax = plt.subplots(figsize=(9, 5))
x = np.arange(len(CLASS_NAMES))
width = 0.35
ax.bar(x - width / 2, [train_counts[c] for c in CLASS_NAMES], width, label="train")
ax.bar(x + width / 2, [test_counts[c] for c in CLASS_NAMES], width, label="test")
ax.set_xticks(x)
ax.set_xticklabels(CLASS_NAMES, rotation=30, ha="right")
ax.set_xlabel("Emotion class")
ax.set_ylabel("Number of images")
ax.set_title("FER2013 class distribution")
ax.legend()
fig.tight_layout()
fig.savefig(FIGURES_DIR / "class_distribution.png", dpi=200)
plt.show()
```

## Sample faces — 5 per class

```python
fig, axes = plt.subplots(len(CLASS_NAMES), 5, figsize=(10, 14))
for row, cls in enumerate(CLASS_NAMES):
    cls_dir = FER_DIR / "train" / cls
    samples = sorted(cls_dir.glob("*"))[:5]
    for col, ax in enumerate(axes[row]):
        ax.axis("off")
        if col < len(samples):
            img = Image.open(samples[col])
            ax.imshow(img, cmap="gray")
        if col == 0:
            ax.text(-0.15, 0.5, cls, transform=ax.transAxes, ha="right", va="center", fontsize=10)
fig.suptitle("FER2013 sample faces (5 per class)")
fig.tight_layout()
fig.savefig(FIGURES_DIR / "sample_grid.png", dpi=200, bbox_inches="tight")
plt.show()
```

## Image size and mode

```python
sample_path = next((FER_DIR / "train" / CLASS_NAMES[0]).glob("*"))
with Image.open(sample_path) as img:
    print("Image size:", img.size, "| mode:", img.mode)
```
